```python
# Outlier detection using the IQR method

def detect_outliers(data, column):
    Q1 = data[column].quantile(0.25)
    Q3 = data[column].quantile(0.75)

    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outliers = data[
        (data[column] < lower_bound) |
        (data[column] > upper_bound)
    ]

    print("First quartile (Q1):", Q1)
    print("Third quartile (Q3):", Q3)
    print("IQR:", IQR)
    print("Lower bound:", lower_bound)
    print("Upper bound:", upper_bound)
    print("Number of potential outliers:", len(outliers))

    return outliers


print("MATHEMATICS: Outlier Analysis")
math_outliers = detect_outliers(math_df, "G3")

print("\nPORTUGUESE: Outlier Analysis")
portuguese_outliers = detect_outliers(portuguese_df, "G3")
```

## Exploratory Data Analysis — Key Findings

### 1. Final Grade Distribution
The Mathematics dataset contains 395 students, while the Portuguese dataset contains 649 students. The mean final grades are approximately 10.42 for Mathematics and 11.91 for Portuguese. The median grades are 11 and 12, respectively. Mathematics grades have greater variability, with a standard deviation of approximately 4.58 compared with 3.23 for Portuguese.

### 2. Study Time and Academic Performance
Students reporting 5–10 hours of weekly study time achieved average Mathematics and Portuguese grades of approximately 11.40 and 13.23, respectively. Students reporting less than two hours achieved averages of approximately 10.05 in Mathematics and 10.84 in Portuguese. The relationship is not perfectly linear, as students reporting more than 10 hours did not achieve the highest average grades in either subject.

### 3. Absences and Final Grades
The correlation between absences and final grades is approximately 0.034 for Mathematics and −0.091 for Portuguese. These values indicate little linear association in Mathematics and a weak negative association in Portuguese. Individual grades vary considerably at different absence levels.

### 4. Parental Education
Higher parental education levels generally correspond to higher average grades in both subjects, particularly for students whose mothers or fathers have education level 4. However, some categories contain very few students, so their averages should be interpreted cautiously.

### 5. Family Background
Students in smaller families (three or fewer members) achieved slightly higher average Mathematics grades than students in larger families. The difference was smaller in Portuguese. Parents' living arrangements showed little difference in average Portuguese grades, while the Mathematics averages differed somewhat.

### 6. Previous Academic Performance
Previous grades have the strongest positive correlations with final grades. The correlation between G2 and G3 is approximately 0.90 in Mathematics and 0.92 in Portuguese. The correlation between G1 and G3 is approximately 0.80 in Mathematics and 0.83 in Portuguese.

### 7. Extracurricular Activities
Students participating in extracurricular activities achieved average grades of approximately 10.49 in Mathematics and 12.10 in Portuguese. Students not participating achieved averages of approximately 10.34 and 11.72, respectively. The differences are relatively small and do not establish a causal relationship.

### 8. Gender Differences
Male students achieved a higher average Mathematics grade (10.91) than female students (9.97). In Portuguese, female students achieved a higher average (12.25) than male students (11.41). These patterns differ by subject.

### 9. Previous Class Failures
Students with no previous class failures achieved the highest average final grades in both subjects. The correlation between previous failures and final grades is approximately −0.36 in Mathematics and −0.39 in Portuguese.

### 10. Outlier Detection
The IQR method identified no final-grade outliers in Mathematics and 16 potential low outliers in Portuguese. These observations should not be removed automatically because they may represent genuine student performance.

### Overall Conclusion
Previous academic grades and previous class failures show stronger relationships with final grades than most of the other variables examined. Study time and parental education show weaker positive associations, while absences have little linear relationship with Mathematics grades and a weak negative relationship with Portuguese grades.

These findings are descriptive. Correlation and group comparisons do not establish causation, and the analysis does not account for every factor that may influence student performance.

## Recommendations

Based on the exploratory data analysis, the following recommendations may help schools improve student academic performance:

1. **Monitor previous academic performance:** G1 and G2 are strongly correlated with final grades. Schools can identify students experiencing difficulties early and provide additional academic support.

2. **Support students with previous failures:** Students with more previous class failures tend to have lower final grades. Individual learning plans, remedial classes, and regular progress reviews may help these students.

3. **Encourage effective study habits:** Schools can help students develop consistent study schedules and effective learning strategies. However, the results suggest that more study time does not always correspond to higher grades.

4. **Consider parental education as contextual information:** Parental education shows some association with student performance. Schools can provide additional learning resources and guidance to students who may need them, without making assumptions about their abilities based on their family background.

5. **Investigate attendance patterns:** Absences have a weak relationship with final grades in these datasets. Schools should examine individual attendance and academic difficulties rather than assuming that absences alone explain performance.

6. **Provide subject-specific support:** Gender-related grade patterns differ between Mathematics and Portuguese. Schools should assess individual learning needs and ensure that all students have equal access to support.

7. **Avoid conclusions based solely on group averages:** Factors such as family size and extracurricular activities show relatively small differences in average grades. Further analysis would be needed to determine their practical significance.

### Recommendation Summary

Schools should prioritize early identification of academic difficulties, monitor progress throughout the school year, and provide targeted support based on individual student needs.
